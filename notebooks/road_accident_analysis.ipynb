## Feature Selection & Target Variable Identification

```python
classification_drop_cols = [
    'Num',
    'Accident_severity',
    'Number_of_casualties',
    'Casualty_class',
    'Sex_of_casualty',
    'Age_band_of_casualty',
    'Casualty_severity',
    'Work_of_casuality',
    'Fitness_of_casuality'
]

X_class = df_clean.drop(
    columns=classification_drop_cols
)

y_class = df_clean['Accident_severity']

print("Classification Features:", X_class.shape)
print("Classification Target:", y_class.shape)
```

```python
regression_drop_cols = [
    'Num',
    'Accident_severity',
    'Number_of_casualties',
    'Casualty_class',
    'Sex_of_casualty',
    'Age_band_of_casualty',
    'Casualty_severity',
    'Work_of_casuality',
    'Fitness_of_casuality'
]

X_reg = df_clean.drop(
    columns=regression_drop_cols
)

y_reg = df_clean['Number_of_casualties']

print("Regression Features:", X_reg.shape)
print("Regression Target:", y_reg.shape)
```

```python
print("Classification Target Distribution:")
print(y_class.value_counts())

print("\nRegression Target Statistics:")
print(y_reg.describe())
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_class_train, X_class_test, y_class_train, y_class_test = train_test_split(
    X_class,
    y_class,
    test_size=0.20,
    random_state=42,
    stratify=y_class
)

print("Classification Training Shape:", X_class_train.shape)
print("Classification Testing Shape:", X_class_test.shape)
```

```python
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)

print("Regression Training Shape:", X_reg_train.shape)
print("Regression Testing Shape:", X_reg_test.shape)
```

```python
print("Original Dataset Distribution:")
print(y_class.value_counts(normalize=True) * 100)

print("\nTraining Dataset Distribution:")
print(y_class_train.value_counts(normalize=True) * 100)

print("\nTesting Dataset Distribution:")
print(y_class_test.value_counts(normalize=True) * 100)
```

```python
print("Missing Values in Classification Training:")
print(X_class_train.isnull().sum().sum())

print("\nMissing Values in Regression Training:")
print(X_reg_train.isnull().sum().sum())
```

## Missing Value Imputation & Feature Encoding

```python
numerical_cols = X_class_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = X_class_train.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Numerical Columns:", numerical_cols)
print("\nTotal Numerical Columns:", len(numerical_cols))

print("\nTotal Categorical Columns:", len(categorical_cols))
print("\nCategorical Columns:", categorical_cols)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
numerical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])
```

```python
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
preprocessor = ColumnTransformer([
    ('numerical', numerical_pipeline, numerical_cols),
    ('categorical', categorical_pipeline, categorical_cols)
])
```

```python
X_class_train_processed = preprocessor.fit_transform(
    X_class_train
)

X_class_test_processed = preprocessor.transform(
    X_class_test
)
```

```python
reg_preprocessor = ColumnTransformer([
    ('numerical', numerical_pipeline, numerical_cols),
    ('categorical', categorical_pipeline, categorical_cols)
])

X_reg_train_processed = reg_preprocessor.fit_transform(
    X_reg_train
)

X_reg_test_processed = reg_preprocessor.transform(
    X_reg_test
)
```

```python
print("Classification Training:", X_class_train_processed.shape)
print("Classification Testing:", X_class_test_processed.shape)

print("\nRegression Training:", X_reg_train_processed.shape)
print("Regression Testing:", X_reg_test_processed.shape)
```

```python
print(
    "Classification Missing Values:",
    np.isnan(X_class_train_processed.data).sum()
    if hasattr(X_class_train_processed, 'data')
    else np.isnan(X_class_train_processed).sum()
)
```

```python
from scipy import sparse

def count_missing(X):
    if sparse.issparse(X):
        return np.isnan(X.data).sum()
    return np.isnan(X).sum()

print("Classification Train:", count_missing(X_class_train_processed))
print("Classification Test:", count_missing(X_class_test_processed))
print("Regression Train:", count_missing(X_reg_train_processed))
print("Regression Test:", count_missing(X_reg_test_processed))
```

## Regression Model: Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

linear_model = LinearRegression()

linear_model.fit(
    X_reg_train_processed,
    y_reg_train
)

y_reg_pred = linear_model.predict(X_reg_test_processed)

mae = mean_absolute_error(y_reg_test, y_reg_pred)
rmse = np.sqrt(mean_squared_error(y_reg_test, y_reg_pred))
r2 = r2_score(y_reg_test, y_reg_pred)

print("Multiple Linear Regression Results")
print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```
