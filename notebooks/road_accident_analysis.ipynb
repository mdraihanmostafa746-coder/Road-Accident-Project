## Data Cleaning & Preprocessing

```python
df_raw = df.copy()

df_clean = df.copy()

print("Original Dataset Shape:", df_raw.shape)
print("Working Dataset Shape:", df_clean.shape)
```

```python
df_clean = df_clean.replace(
    r'(?i)^\s*na\s*$',
    np.nan,
    regex=True
)
```

```python
print("Missing Values After Conversion:")
print(df_clean.isnull().sum().sum())

print("\nMissing Values Per Column:")
print(df_clean.isnull().sum().sort_values(ascending=False))
```

```python
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
missing_summary = pd.DataFrame({
    'Missing_Count': df_clean.isnull().sum(),
    'Missing_Percentage': df_clean.isnull().mean() * 100
})

missing_summary = missing_summary.sort_values(
    by='Missing_Percentage',
    ascending=False
)

missing_summary
```

```python
plt.figure(figsize=(14, 7))

sns.heatmap(
    df_clean.isnull(),
    cbar=False,
    yticklabels=False
)

plt.title("Missing Values in Road Accident Dataset")
plt.xlabel("Columns")
plt.show()
```

## Feature Selection & Target Variable Identification

```python
classification_drop_cols = [
    'Num',
    'Accident_severity',
    'Number_of_casualties',
    'Casualty_class',
    'Sex_of_casualty',
    'Age_band_of_casualty',
    'Casualty_severity',
    'Work_of_casuality',
    'Fitness_of_casuality'
]

X_class = df_clean.drop(
    columns=classification_drop_cols
)

y_class = df_clean['Accident_severity']

print("Classification Features:", X_class.shape)
print("Classification Target:", y_class.shape)
```

```python
regression_drop_cols = [
    'Num',
    'Accident_severity',
    'Number_of_casualties',
    'Casualty_class',
    'Sex_of_casualty',
    'Age_band_of_casualty',
    'Casualty_severity',
    'Work_of_casuality',
    'Fitness_of_casuality'
]

X_reg = df_clean.drop(
    columns=regression_drop_cols
)

y_reg = df_clean['Number_of_casualties']

print("Regression Features:", X_reg.shape)
print("Regression Target:", y_reg.shape)
```

```python
print("Classification Target Distribution:")
print(y_class.value_counts())

print("\nRegression Target Statistics:")
print(y_reg.describe())
```
